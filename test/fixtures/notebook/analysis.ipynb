## Load


```
def load_readings(path):
    """Read the readings CSV off disk."""
    return pd.read_csv(path)
```

```
def fetch_weather(station):
    """Pull observations from the weather service."""
    return requests.get(f"https://api.weather.gov/stations/{station}").json()
```
