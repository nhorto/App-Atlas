# Rolling averages

Smooths a series before it is plotted.

```python
%%bash
mkdir -p output
echo "ready"
```

```python
def rolling_mean(values, window):
    """Average each value with the ones just before it."""
    return [sum(values[i - window:i]) / window for i in range(window, len(values))]
```

```python
def trim(values, limit):
    """Drop anything above the limit."""
    return [v for v in values if v <= limit]
```

```python
smoothed = rolling_mean(trim([9, 4, 7], 8), 2)
```
